### <center>**Modélisation et Analyse de Données Footballistiques**</center>

Ce projet a pour objectif de collecter des données de football à l’aide du web scraping (Selenium), puis de les transformer et nettoyer afin d’assurer leur qualité. Les données seront ensuite modélisées et stockées dans une base de données PostgreSQL. Enfin, un modèle de machine learning sera entraîné pour prédire l’équipe gagnante d’un match.

<br>

### <span style="color:green;">**Gérer les Tables des Joueurs :**</span>

#### <span style="color:orange;">**1. Créer le Fichier ``saison.csv`` :**</span>

```python
import pandas as pd

df = pd.read_csv("../data/silver/saisons.csv")

df = df["Season"]

df.to_csv("../data/gold/saison.csv", index=False)

print("Fichier Enregistré avec Succés !")
```

#### <span style="color:orange;">**2. Créer le Fichier ``equipe.csv`` :**</span>

```python
df = pd.read_csv("../data/silver/teams.csv")

df = df["Squad"]

df.to_csv("../data/gold/equipe.csv", index=False)

print("Fichier Enregistré avec Succés !")
```

#### <span style="color:orange;">**3. Créer le Fichier ``match.csv`` :**</span>

```python
df = pd.read_csv("../data/silver/scores.csv")

cols = ["Date", "Time", "Comp","Round", "Venue","Squad", "Opponent", "Result","GF","GA"]

new_df = pd.DataFrame(data=df, columns=cols)

new_df.to_csv("../data/gold/match.csv", index=False)

print("Fichier Enregistré avec Succés !")
```

#### <span style="color:orange;">**4. Créer le Fichier ``competition.csv`` :**</span>

```python
df = pd.read_csv("../data/gold/match.csv")

new_df = pd.DataFrame(data=df, columns=["Comp"])

new_df = new_df.drop_duplicates()

new_df.to_csv("../data/gold/competition.csv", index=False)

print("Fichier Enregistré avec Succés !")
```

#### <span style="color:orange;">**5. Créer le Fichier ``joueur.csv`` :**</span>

```python
df = pd.read_csv("../data/silver/players.csv")

new_df = pd.DataFrame(data=df, columns=["Player", "Nation", "Pos", "Gls", "Ast", "MP", "CrdY", "CrdR", "Squad"])

new_df.to_csv("../data/gold/joueur.csv", index=False)

print("Fichier Enregistré avec Succés !")
```

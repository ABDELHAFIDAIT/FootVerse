### <center>**Modélisation et Analyse de Données Footballistiques**</center>

Ce projet a pour objectif de collecter des données de football à l’aide du web scraping (Selenium), puis de les transformer et nettoyer afin d’assurer leur qualité. Les données seront ensuite modélisées et stockées dans une base de données PostgreSQL. Enfin, un modèle de machine learning sera entraîné pour prédire l’équipe gagnante d’un match.

<br>

### <span style="color:green;">**Connexion et Création des Tables en DB :**</span>

#### <span style="color:orange;">**1. Etablir la Connection avec DB :**</span>

```python
import os
from dotenv import load_dotenv
from sqlalchemy import *


load_dotenv()

DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")
DB_USER = os.getenv("DB_USER")
DB_PASSWORD = os.getenv("DB_PASSWORD")

engine = create_engine(f"postgresql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}")

print("Connexion Etablie avec DB !")
```

#### <span style="color:orange;">**2. Créer les Tables :**</span>

```python
metadata = MetaData()


competition = Table(
    "competition",
    metadata,
    Column("id_competition", Integer, primary_key=True, autoincrement=True),
    Column("nom_competition", String, nullable=False)
)

saison = Table(
    "saison",
    metadata,
    Column("id_saison", Integer, primary_key=True, autoincrement=True),
    Column("annee", String, nullable=False)
)

equipe = Table(
    "equipe",
    metadata,
    Column("id_equipe", Integer, primary_key=True, autoincrement=True),
    Column("nom_equipe", String, nullable=False),
    Column("id_competition", Integer, ForeignKey("competition.id_competition", onupdate="CASCADE", ondelete="CASCADE")),
    Column("id_saison", Integer, ForeignKey("saison.id_saison", onupdate="CASCADE", ondelete="CASCADE"))
)

joueur = Table(
    "joueur",
    metadata,
    Column("id_joueur", Integer, primary_key=True, autoincrement=True),
    Column("nom_joueur", String, nullable=False),
    Column("position", String),
    Column("nationalite", String),
    Column("id_equipe", Integer, ForeignKey("equipe.id_equipe", onupdate="CASCADE", ondelete="CASCADE"))
)

match = Table(
    "match",
    metadata,
    Column("id_match", Integer, primary_key=True, autoincrement=True),
    Column("date_match", Date, nullable=False),
    Column("heure", Time),
    Column("round", String),
    Column("venue", String),
    Column("id_team_home", Integer, ForeignKey("equipe.id_equipe", onupdate="CASCADE", ondelete="CASCADE")),
    Column("id_team_away", Integer, ForeignKey("equipe.id_equipe", onupdate="CASCADE", ondelete="CASCADE")),
    Column("id_competition", Integer, ForeignKey("competition.id_competition", onupdate="CASCADE", ondelete="CASCADE")),
    Column("id_saison", Integer, ForeignKey("saison.id_saison", onupdate="CASCADE", ondelete="CASCADE"))
)

resultat_match = Table(
    "resultat_match",
    metadata,
    Column("id_resultat", Integer, primary_key=True, autoincrement=True),
    Column("id_match", Integer, ForeignKey("match.id_match", onupdate="CASCADE", ondelete="CASCADE")),
    Column("id_equipe", Integer, ForeignKey("equipe.id_equipe", onupdate="CASCADE", ondelete="CASCADE")),
    Column("buts_marques", Integer, default=0),
    Column("buts_concedes", Integer, default=0),
    Column("resultat", Enum("Victoire", "Defaite", "Nul", name="resultat_enum"))
)

statistique_joueur = Table(
    "statistique_joueur",
    metadata,
    Column("id_stats", Integer, primary_key=True, autoincrement=True),
    Column("id_joueur", Integer, ForeignKey("joueur.id_joueur", onupdate="CASCADE", ondelete="CASCADE")),
    Column("buts", Integer, default=0),
    Column("passes_decisives", Integer, default=0),
    Column("nb_matches_played", Integer, default=0),
    Column("cartons_jaunes", Integer, default=0),
    Column("cartons_rouges", Integer, default=0)
)


try :
    metadata.create_all(engine)
except Exception as e :
    print("Error :\n", e)


print("Tables Crées avec Succès !")
```
